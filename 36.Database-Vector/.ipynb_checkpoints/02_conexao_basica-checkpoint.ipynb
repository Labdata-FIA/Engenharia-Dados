# Aula 2 — Conexão Básica com PostgreSQLNeste notebook vamos abrir uma conexão com o banco, habilitar a extensão**pgvector** e conferir se tudo ficou funcionando.Requisitos: `psycopg2-binary`, `python-dotenv` e um PostgreSQL rodando.

## 1. Importações e configuração

```python
import osimport psycopg2from dotenv import load_dotenvload_dotenv()# String de conexão (pode vir do .env)STRING_CONEXAO = os.getenv(    "DATABASE_URL", "postgresql://postgres:postgres@localhost:5432/postgres")print(f"Servidor configurado: {STRING_CONEXAO.split('@')[-1]}")
```

## 2. Função de conexão

```python
def abrir_conexao():    """Abre a conexão com o PostgreSQL e devolve o objeto de conexão."""    ligacao = psycopg2.connect(STRING_CONEXAO)    ligacao.autocommit = True    return ligacao
```

## 3. Habilitando a extensão pgvector

```python
def habilitar_pgvector(ligacao):    """Cria a extensão vector, caso ainda não exista."""    ponteiro = ligacao.cursor()    ponteiro.execute("CREATE EXTENSION IF NOT EXISTS vector;")    print("✅ Extensão pgvector habilitada")    ponteiro.close()
```

## 4. Conferindo a instalação

```python
def conferir_pgvector(ligacao):    """Verifica a versão da extensão e testa o tipo vector."""    ponteiro = ligacao.cursor()    # Consulta o catálogo de extensões    ponteiro.execute("SELECT * FROM pg_extension WHERE extname = 'vector';")    registro = ponteiro.fetchone()    if registro:        print(f"✅ Versão do pgvector: {registro[4]}")    else:        print("❌ pgvector não foi encontrado")    # Testa a conversão para o tipo vector    ponteiro.execute("SELECT '[10,20,30]'::vector;")    vetor_teste = ponteiro.fetchone()[0]    print(f"✅ Teste de vetor: {vetor_teste}")    ponteiro.close()
```

## 5. Executando tudo

```python
print("Conectando ao PostgreSQL...")ligacao = abrir_conexao()print("✅ Conectado!")print("\nHabilitando o pgvector...")habilitar_pgvector(ligacao)print("\nConferindo a instalação...")conferir_pgvector(ligacao)
```

## 6. Encerrando a conexão

```python
ligacao.close()print("✅ Conexão encerrada. Configuração concluída!")
```
