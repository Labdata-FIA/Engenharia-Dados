# Aula 2 — Conexão Básica com PostgreSQLNeste notebook vamos abrir uma conexão com o banco, habilitar a extensão**pgvector** e conferir se tudo ficou funcionando.Requisitos: `psycopg2-binary`, `python-dotenv` e um PostgreSQL rodando.

## 1. Importações e configuração

```python
import psycopg2
# String de conexão (pode vir do .env)
DATABASE_URL = "postgresql://postgres:postgres@pgvector:5432/postgres"

print(f"Servidor configurado: {DATABASE_URL.split('@')[-1]}")
```

## 2. Função de conexão

```python
def connect():
    """Abre a conexão com o PostgreSQL e devolve o objeto de conexão"""
    conn = psycopg2.connect(DATABASE_URL)
    conn.autocommit = True
    return conn
```

## 3. Habilitando a extensão pgvector

```python
def enable_pgvector(conn):
    """Cria a extensão vector, caso ainda não exista."""
    cur = conn.cursor()
    cur.execute("CREATE EXTENSION IF NOT EXISTS vector;")
    print("✅ pgvector extension enabled")
    cur.close()
```

## 4. Conferindo a instalação

```python
def verify_pgvector(conn):
    """Verifica a versão da extensão e testa o tipo vector."""
    cur = conn.cursor()

    # Check extension
    cur.execute("SELECT * FROM pg_extension WHERE extname = 'vector';")
    result = cur.fetchone()

    if result:
        print(f"✅ pgvector version: {result[4]}")
    else:
        print("❌ pgvector not installed")

    # Test vector type
    cur.execute("SELECT '[1,2,3]'::vector;")
    vector = cur.fetchone()[0]
    print(f"✅ Teste de vetort: {vector}")

    cur.close()
```

## 5. Executando tudo

```python
print("Connecting to PostgreSQL...")
conn = connect()
print("✅ Connected!")

print("\nEnabling pgvector...")
enable_pgvector(conn)

print("\nVerifying installation...")
verify_pgvector(conn)

conn.close()
print("\n✅ Setup complete!")
```
